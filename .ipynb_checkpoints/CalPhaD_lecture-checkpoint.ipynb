# 1. Зачем нужен CALPHAD: что мы вообще считаем

**Цель проекта:** по заданным температуре $T$ и составу сплава найти **равновесное состояние** —
какие фазы присутствуют, в каких количествах и с какими составами — используя термодинамическую
базу данных в формате TDB.

**Физический принцип.** При постоянных $T$, давлении $P$ и числе молей компонентов $N_i$ равновесие —
это состояние с **минимумом полной энергии Гиббса**

$$G = \sum_{\alpha} n^{\alpha}\, G_m^{\alpha}(T, \mathbf{y}^{\alpha}) \;\to\; \min,
\qquad \sum_{\alpha} n^{\alpha} x_i^{\alpha} = N_i ,$$

где $\alpha$ — номер фазы, $n^{\alpha}$ — её количество, $G_m^{\alpha}$ — энергия Гиббса фазы,
$x_i^{\alpha}$ — мольные доли.

**Идея CALPHAD.** Энергию Гиббса *каждой фазы* описывают аналитической моделью с параметрами,
подогнанными под эксперимент. Параметры лежат в TDB-файле. Тогда расчёт равновесия = «взять модели
фаз из файла + решить задачу минимизации». Наш проект делится ровно на эти две части:

| Часть | Что делает | Статус |
|---|---|---|
| **A. Модель** | читает TDB и вычисляет $G_m(T,\mathbf{y})$ каждой фазы | почти готова (нет магнитного вклада) |
| **B. Солвер** | ищет минимум $G$ по фазам и составам | ещё не начат |

## Подрешёточная модель (CEF)

Фаза описывается **подрешётками**. Пример: аустенит `FCC_A1` в нашей базе — две подрешётки
$(\text{Fe, Cr, Ni, ...})_1(\text{C, N, VA, ...})_1$: в первой — металлы, во второй — внедрённые атомы
или **вакансии** `VA`. Состав фазы задают **site fractions** $y_{s,i}$ — доля вида $i$ в подрешётке $s$,
причём в каждой подрешётке $\sum_i y_{s,i}=1$.

Энергия Гиббса фазы на формульную единицу:

$$G_m = \underbrace{G^{ref}}_{\text{референс}} + \underbrace{G^{id}}_{\text{идеальное смешение}}
+ \underbrace{G^{ex}}_{\text{избыточная}} + \underbrace{G^{mag}}_{\text{магнитная}}$$

* **Reference**: $G^{ref}=\sum_I \Big(\prod_s y_{s,I_s}\Big)\,{}^0G_I$ — сумма энергий «крайних членов»
  (end-members: по одному виду в каждой подрешётке), взвешенная произведением site fractions.
* **Ideal**: $G^{id}=RT\sum_s a_s\sum_i y_{s,i}\ln y_{s,i}$, где $a_s$ — стехиометрия подрешётки (site ratio).
  Вакансия `VA` участвует как обычный вид.
* **Excess**: взаимодействия между видами — полиномы Редлиха–Кистера (бинарные), Мугьяну (тройные) и т. д. по $L$-параметрам.
* **Magnetic**: модель Индена–Хиллерта–Ярла (нужна для Fe, Co, Ni, Mn, Cr и сплавов с ними) — **следующий шаг проекта**.

Состав в привычных мольных долях элементов $x_i$ получается из $y$ пересчётом (с учётом того, что `VA` — не атом).

# 2. Формат TDB: из чего состоит файл

TDB — текстовый файл. Каждая «команда» заканчивается `!`, строки с `$` — комментарии. Основные команды:

| Команда | Смысл | Наш парсер |
|---|---|---|
| `ELEMENT` | элемент, референсное состояние, масса, $H_{298}$, $S_{298}$ | `ElementTdbParser` |
| `TYPE_DEFINITION` | дополнительная модель фазы (у нас — только `MAGNETIC`) | `TypeDefinitionTdbParser` |
| `PHASE` | имя фазы, число подрешёток, site ratios, ссылка на `TYPE_DEFINITION` | `PhaseTdbParser` |
| `CONSTITUENT` | какие виды допустимы в каждой подрешётке | `PhaseTdbParser` |
| `FUNCTION` | именованная функция $G(T)$ (например `GHSERFE`) | `SerTdbParser`, `OtherThanSerTdbParser`, `AutoFunctionTdbParser` |
| `PARAMETER` | параметры модели фазы: `G`, `L`, `TC`, `BMAGN` | `ParameterTdbParser` |

Посмотрим на настоящие строки из вашей базы:

````python
raw = open(DB_PATH, encoding='utf-8').read()
lines = raw.split('\n')

def show_from(predicate, n=1, title=''):
    for i, line in enumerate(lines):
        if predicate(line.strip()):
            print(title or '--- строка', i + 1)
            print('\n'.join(l.rstrip() for l in lines[i:i + n]))
            print()
            return
    print('не найдено')

show_from(lambda s: s.split()[:2] == ['ELEMENT', 'FE'], 1, '[ELEMENT]')
show_from(lambda s: s.startswith('TYPE_DEFINITION') and 'FCC_A1' in s, 1, '[TYPE_DEFINITION]')
show_from(lambda s: s.startswith('PHASE FCC_A1'), 4, '[PHASE]')
show_from(lambda s: s.startswith('CONSTITUENT FCC_A1') or s.startswith('CONST FCC_A1'), 3, '[CONSTITUENT]')
show_from(lambda s: s.startswith('FUNCTION GHSERFE'), 9, '[FUNCTION]')
show_from(lambda s: s.startswith('PARAMETER G(FCC_A1,FE:VA;0)'), 4, '[PARAMETER G]')
show_from(lambda s: s.startswith('PARAMETER L(FCC_A1,FE,NI:VA;0)'), 3, '[PARAMETER L]')
````

Обратите внимание на три важные особенности формата, которые определили код парсеров:

1. **Выражения идут диапазонами по температуре**: `T_min  выражение; T_max  Y|N`. Метка `Y` — «за этим диапазоном
   есть следующий», `N` — «последний». Первый диапазон начинается с явного `T_min` (обычно 273 К), следующие — с `T_max` предыдущего.
2. **Ссылки на другие функции** записываются как `GHSERFE#`. Значит, функции зависят друг от друга, и вычислять их надо рекурсивно.
3. **Заголовок параметра** `G(FCC_A1,FE:VA;0)` кодирует всё сразу: тип (`G`), фазу, виды по подрешёткам
   (`:` разделяет подрешётки, `,` — виды внутри подрешётки) и порядок `;0` члена Редлиха–Кистера.

# 5. Слой 2: функции Гиббса (полиномы по температуре)

## 5.1 Полином SGTE

Почти все термодинамические функции в TDB — это полиномы по $T$ в стандартной форме SGTE:

$$G(T) = b + t_1 T + t_{lnt}\, T\ln T + t_2 T^2 + t_3 T^3 + t_4 T^4 + t_7 T^7
+ \frac{t_{m1}}{T} + \frac{t_{m2}}{T^2} + \frac{t_{m3}}{T^3} + \frac{t_{m9}}{T^9}
\;+\; \sum_k c_k\, G_k(T)$$

Последняя сумма — **ссылки на другие функции** (`GHSERFE#`) с коэффициентами $c_k$.
Каждый температурный диапазон хранится объектом `TdbFunctionData`: границы `t_min`/`t_max`, 11 коэффициентов,
список `dependencies` и коэффициенты `ref_coef`.

## 5.2 Как работает разбор (`TdbParser`)

1. Текст блока режется по ключевому слову `FUNCTION` — получаем выражения по одной функции.
2. Внутри функции выражение режется по границам `; Tmax Y|N` — по одному куску на диапазон.
3. Из куска регулярным выражением достаются **зависимости** (`коэффициент*ИМЯ#`) и **вырезаются** из текста.
4. Оставшееся — сумма слагаемых вида `число * T**n` — раскладывается по словарю `{'': b, '*T': t1, '*T*LN(T)': tlnt, ...}`.

Посмотрим на результат для `GHSERFE` (стандартное состояние железа — ОЦК):

````python
COEFS = ['b', 't1', 'tlnt', 't2', 't3', 't4', 't7', 'tm1', 'tm2', 'tm3', 'tm9']

def nonzero_coefs(f):
    return {k: getattr(f, k) for k in COEFS if getattr(f, k) != 0.0}

for rng in db.functions['GHSERFE']:
    print(f"[{rng.t_min:7.2f} .. {rng.t_max:7.2f}] К  зависимости: {rng.dependencies}")
    print("    ", nonzero_coefs(rng))
````

Проверим руками: подставим коэффициенты в формулу и сравним с `db.evaluate_function` (при $T=1000$ К):

````python
from GibbsEnergyFunctionsDataClasses import TdbFunctionManipulation

T = 1000.0
f = TdbFunctionManipulation.get_function_from_dict('GHSERFE', T, db.functions)   # нужный температурный диапазон
manual = (f.b + f.t1*T + f.tlnt*T*np.log(T) + f.t2*T**2 + f.t3*T**3 + f.t4*T**4 + f.t7*T**7
          + f.tm1/T + f.tm2/T**2 + f.tm3/T**3 + f.tm9/T**9)

print("вручную          :", manual)
print("db.evaluate_function:", db.evaluate_function('GHSERFE', T))
````

## 5.3 Зависимости между функциями и исправленная ошибка

Многие функции выражены через другие. Например, ГЦК-железо — это ОЦК-железо плюс поправка:
`GFEFCC = -1462.4 + 8.282*T + ... + GHSERFE#`. Вычисление **рекурсивное**: чтобы найти `GFEFCC`, сначала считаем `GHSERFE`.

````python
print("GFEFCC:")
for rng in db.functions['GFEFCC']:
    print(f"  [{rng.t_min:7.2f} .. {rng.t_max:7.2f}] К  deps={rng.dependencies} coef={rng.ref_coef}")
    print("      ", nonzero_coefs(rng))

def poly(f, T):
    # полиномиальная часть SGTE (без ссылок на другие функции)
    return (f.b + f.t1*T + f.tlnt*T*np.log(T) + f.t2*T**2 + f.t3*T**3 + f.t4*T**4 + f.t7*T**7
            + f.tm1/T + f.tm2/T**2 + f.tm3/T**3 + f.tm9/T**9)

T = 1000.0
g = TdbFunctionManipulation.get_function_from_dict('GFEFCC', T, db.functions)
print("\nполином GFEFCC без ссылок :", poly(g, T))
print("GFEFCC(1000 К) вручную    :", poly(g, T) + g.ref_coef[0] * db.evaluate_function('GHSERFE', T))
print("GFEFCC(1000 К) из базы    :", db.evaluate_function('GFEFCC', T))

print("\nФункция с ДВУМЯ зависимостями — GTI2NI:")
g2 = TdbFunctionManipulation.get_function_from_dict('GTI2NI', T, db.functions)
print("  зависимости:", g2.dependencies, " коэффициенты:", g2.ref_coef)
manual2 = poly(g2, T) + sum(c * db.evaluate_function(d, T) for c, d in zip(g2.ref_coef, g2.dependencies))
print("  вручную :", manual2)
print("  из базы :", db.evaluate_function('GTI2NI', T))
````

> **Исправленная ошибка.** Старый `_evaluate_data` передавал зависимости в `calculateGibbsEnergy` **одним списком**, а метод принимает
> `*extras` (отдельные аргументы). Проверка `len(extras) != len(dependencies)` проходила только для функций ровно с одной зависимостью,
> а на `GTI2NI` (две зависимости) падала с «Missing required dependency». Теперь:
>
> * `polynome_plus` принимает `*extras`;
> * добавлен **чистый** метод `TdbFunctionData.evaluate(T, dep_values)` — он не пишет `.T`/`.G` в объект и не проверяет диапазон;
>   его и использует `ThermodynamicDatabase` — это быстрый путь для солвера;
> * значения функций кэшируются по `(имя, T)`: при фиксированной $T$ одни и те же `GHSER*` вызываются сотни раз.

Графики: `GHSERFE(T)` и «стабильность решётки» ГЦК относительно ОЦК $G_{FCC}-G_{BCC}$ без магнитного вклада:

````python
Ts = np.arange(300.0, 2400.0, 5.0)
g_bcc = np.array([db.evaluate_function('GHSERFE', T) for T in Ts])
g_fcc = np.array([db.evaluate_function('GFEFCC', T) for T in Ts])

fig, ax = plt.subplots(1, 2, figsize=(12, 4.3))
ax[0].plot(Ts, g_bcc / 1000)
ax[0].axvline(1811, color='gray', ls=':'); ax[0].text(1840, -40, 'граница\nдиапазонов\n1811 К', fontsize=9, va='top')
ax[0].set(title='GHSERFE(T) — Fe, ОЦК', xlabel='T, К', ylabel='G, кДж/моль')
ax[1].plot(Ts, g_fcc - g_bcc, color='C1')
ax[1].axhline(0, color='k', lw=0.8)
ax[1].set(title='GFEFCC − GHSERFE (без магнитного вклада)', xlabel='T, К', ylabel='ΔG, Дж/моль')
plt.tight_layout(); plt.show()
````

Эта разность (чистая решёточная часть, без магнетизма) пересекает ноль **один раз**, около 1800 К.
Настоящее железо ведёт себя иначе: $\alpha\to\gamma$ при ≈1184 К и $\gamma\to\delta$ при ≈1667 К. Разницу создаёт магнитный вклад —

## 5.4 Чистое вычисление и кэш

## 8.1 Reference-член и end-members

$$G^{ref}=\sum_I \Big(\prod_s y_{s,I_s}\Big)\,{}^0G_I$$

Для `FCC_A1` end-member — пара «металл : внедрённый атом/вакансия», например `FE:VA` (чистое ГЦК-железо) или `FE:C` (гипотетический FeC).
`PhaseModel` для каждой такой пары находит параметр `G(FCC_A1,A:B;0)` и считает его значение при $T$.
Проверка руками для $y=(\text{Fe}:\ 0.5\,\text{C}+0.5\,\text{VA})$ при 1200 К:

````python
T = 1200.0
y = m.y_from_sublattices([{'FE': 1.0}, {'C': 0.5, 'VA': 0.5}])

g_fe_va = db.evaluate_parameter(db.find_parameter('FCC_A1', 'G', [['FE'], ['VA']], 0), T)
g_fe_c  = db.evaluate_parameter(db.find_parameter('FCC_A1', 'G', [['FE'], ['C']], 0), T)

print("G(Fe:VA) =", round(g_fe_va, 1), " G(Fe:C) =", round(g_fe_c, 1))
print("вручную 0.5*G(Fe:VA) + 0.5*G(Fe:C) =", 0.5 * g_fe_va + 0.5 * g_fe_c)
print("PhaseModel.gibbs_reference         =", m.gibbs_reference(T, y))
````

**Отсутствующие end-members.** Если для пары нет параметра `G`, `PhaseModel` по умолчанию подставляет $G=0$

````python
print("вручную    :", -R_GAS * T * np.log(2))
print("PhaseModel :", m.gibbs_ideal(T, y))
````

## 8.3 Избыточная энергия: Редлих–Кистер, Мугьяну, «звёздочки»

Всё, что связано с взаимодействиями, собрано в классе **`InteractionSet`** — он превращает список $L$-параметров фазы в массивы NumPy для быстрого расчёта.
Каждый член имеет вид (произведение site fractions) × (множитель Редлиха–Кистера) × (значение параметра при $T$):

* **бинарное** взаимодействие в одной подрешётке: $\;y_i y_j \sum_v {}^vL_{ij}\,(y_i-y_j)^v$ (виды упорядочены по алфавиту — как требует TDB);
* **тройное** (Мугьяну): $\;y_i y_j y_k\,(v_i\,{}^0L + v_j\,{}^1L + v_k\,{}^2L)$, где $v_i = y_i + \tfrac{1-y_i-y_j-y_k}{3}$;
* **четверные и выше**: константа × произведение $y$;
* **реципрокные** $(A,B:C,D)$: $\;y_A y_B y_C y_D\,L$ (для порядков 1 и 2 — множители $(y_A-y_B)$ и $(y_C-y_D)$);
* **`*` («любой вид»)**: так как $\sum_i y_{s,i}=1$, подрешётка со звёздочкой просто выпадает из произведения;
* в TDB встречается взаимодействие, записанное как `G(phase,A,B;n)` вместо `L` — оно тоже учитывается.

### Пример: бинарная система Fe–Ni в ГЦК

Возьмём параметры $L(\text{FCC\_A1},\text{FE,NI}:\text{VA};v)$, вычислим избыточную энергию вручную и сравним с моделью:

````python
def rk_values(db, phase, cons, T, ptype='L', max_order=8):
    # {порядок: значение параметра при T} для всех найденных порядков
    out = {}
    for order in range(max_order + 1):
        r = db.find_parameter(phase, ptype, cons, order)
        if r is not None:
            out[order] = db.evaluate_parameter(r, T)
    return out

T = 1200.0
L_FeNi = rk_values(db, 'FCC_A1', [['FE', 'NI'], ['VA']], T)
print("L(FCC_A1,FE,NI:VA;v) при 1200 К:", {k: round(v, 1) for k, v in L_FeNi.items()})

def y_FeNi(x_ni):
    return m.y_from_sublattices([{'FE': 1 - x_ni, 'NI': x_ni}, {'VA': 1.0}])

x = 0.3
y_fe, y_ni = 1 - x, x
manual = y_fe * y_ni * sum(L * (y_fe - y_ni) ** v for v, L in L_FeNi.items())
print(f"\nx(Ni) = {x}:  вручную {manual:.4f}   PhaseModel.gibbs_excess {m.gibbs_excess(T, y_FeNi(x)):.4f}")
````

Теперь разложим **энергию смешения** $\Delta G_{mix}(x)= G_m(x) - [(1-x)\,G_{Fe}+x\,G_{Ni}]$ на идеальную и избыточную части.
Reference-часть для бинарной системы линейна по $x$, поэтому $\Delta G_{mix}=G^{id}+G^{ex}$ — это тоже проверка:

````python
xs = np.linspace(0.002, 0.998, 300)
g_Fe = m.gibbs(T, y_FeNi(0.0))      # на вершине ideal = excess = 0, остаётся только reference
g_Ni = m.gibbs(T, y_FeNi(1.0))

G_all   = np.array([m.gibbs(T, y_FeNi(x)) for x in xs])
G_ideal = np.array([m.gibbs_ideal(T, y_FeNi(x)) for x in xs])
G_exc   = np.array([m.gibbs_excess(T, y_FeNi(x)) for x in xs])
dG_mix  = G_all - ((1 - xs) * g_Fe + xs * g_Ni)

print("max |ΔG_mix − (ideal + excess)| =", np.max(np.abs(dG_mix - (G_ideal + G_exc))), "Дж/моль")

plt.plot(xs, dG_mix, lw=2.5, label='ΔG смешения (всё)')
plt.plot(xs, G_ideal, '--', label='идеальная часть')
plt.plot(xs, G_exc, '--', label='избыточная часть')
plt.xlabel('x(Ni)'); plt.ylabel('Дж/моль'); plt.title('ГЦК Fe–Ni, 1200 К (без магнитного вклада!)'); plt.legend(); plt.show()
````

Идеальная часть всегда отрицательна и симметрична (она «заставляет» элементы смешиваться), избыточная — отражает химию системы.
**Предупреждение:** для Fe–Ni магнитный вклад очень велик, поэтому эта кривая пока физически неполна.

### Пример: тройная система Cr–Fe–Ni (Мугьяну)

Здесь в одной подрешётке три вида. В базе для тройки заданы порядки 0, 1, 2. Проверим сумму: три бинарные пары + тройной член.

````python
T = 1200.0
yv = {'CR': 0.18, 'FE': 0.70, 'NI': 0.12}                      # сумма = 1
y_ter = m.y_from_sublattices([yv, {'VA': 1.0}])

def binary_part(a, b):
    a, b = sorted((a, b))
    Ls = rk_values(db, 'FCC_A1', [[a, b], ['VA']], T)
    return yv[a] * yv[b] * sum(L * (yv[a] - yv[b]) ** v for v, L in Ls.items())

bin_total = binary_part('CR', 'FE') + binary_part('CR', 'NI') + binary_part('FE', 'NI')

Lt = rk_values(db, 'FCC_A1', [['CR', 'FE', 'NI'], ['VA']], T)     # порядки 0..2, виды по алфавиту
print("тройные параметры (порядок: значение):", {k: round(v, 1) for k, v in Lt.items()})
Ls3 = [Lt.get(0, 0.0), Lt.get(1, 0.0), Lt.get(2, 0.0)]
if set(Lt) == {0}:                                           # условность: единственный L0 берётся для всех трёх
    Ls3 = [Lt[0]] * 3
ys = [yv['CR'], yv['FE'], yv['NI']]
v_i = [yi + (1 - sum(ys)) / 3 for yi in ys]
ternary = ys[0] * ys[1] * ys[2] * sum(vi * Li for vi, Li in zip(v_i, Ls3))

print("\nбинарные пары :", round(bin_total, 3))
print("тройной член  :", round(ternary, 3))
print("вручную всего :", round(bin_total + ternary, 3))
print("PhaseModel    :", round(m.gibbs_excess(T, y_ter), 3))
````

Совпадение говорит о том, что векторизованная реализация согласована с формулами. Теперь посмотрим на «экзотику» из базы, которую `InteractionSet` обязан был обработать:

````python
print("L-параметры со звёздочкой (любой вид подрешётки):")
for ph_name, by_type in db.parameters.items():
    for p in by_type['L']:
        if any('*' in g for g in p.constituents):
            print(f"   {p.name}")
            break

print("\nреципрокное взаимодействие (A,B:C,D) с порядками 1 и 2 — ГЦК, Fe–Ti : C–VA:")
for p in db.parameters['FCC_A1']['L']:
    if p.constituents == [['FE', 'TI'], ['C', 'VA']]:
        print(f"   {p.name}:  {p.b:+.1f} {p.t1:+.2f}*T")

im = PhaseModel(db, 'FCC_A1').excess
print(f"\nФаза FCC_A1: простых членов {im.n_plain}, тройных {im.n_tern}, нереализованных {len(im.unsupported)}")
````

## 8.4 Градиент $\partial G/\partial y_{s,i}$

Солверу нужны не только значения $G$, но и **производные**: условия равновесия — это равенство химических потенциалов,
а они выражаются через $\partial G/\partial y$. Мы реализовали аналитический градиент всех трёх частей (reference, ideal, excess).
Проверим его конечными разностями:

````python
rng = np.random.default_rng(1)
y = np.zeros(m.n_y)
for s in range(m.n_sub):                                 # случайные y, сумма в каждой подрешётке = 1
    sl = slice(m._offsets[s], m._offsets[s + 1])
    v = rng.random(sl.stop - sl.start) + 0.05
    y[sl] = v / v.sum()

T = 1200.0
g = m.gradient(T, y)
h = 1e-6
print(f"{'индекс':>7} {'вид':>10} {'аналитич.':>16} {'конечн. разн.':>16}")
for i in [0, 4, 10, 21, 25]:
    yp, ym = y.copy(), y.copy(); yp[i] += h; ym[i] -= h
    fd = (m.gibbs(T, yp) - m.gibbs(T, ym)) / (2 * h)
    s, sp = m.species[i]
    print(f"{i:>7} {('s%d:%s' % (s, sp)):>10} {g[i]:>16.4f} {fd:>16.4f}")
````

## 8.5 Нормировка на моль атомов и идея химического потенциала

`PhaseModel.gibbs` возвращает энергию **на формульную единицу** (так, как задана модель в TDB: $a_1=a_2=1$ для `FCC_A1`).
Но в равновесии мы сравниваем фазы при одинаковом числе **атомов**, а вакансии атомами не считаются. Поэтому:

$$N_{at} = \sum_s a_s \sum_{i\neq VA} y_{s,i}, \qquad G_m^{(at)}=\frac{G_m}{N_{at}},
\qquad x_i = \frac{\sum_s a_s\,y_{s,i}}{N_{at}} .$$

Для аустенита Fe–C: $N_{at}=1+y_C$ и $x_C = y_C/(1+y_C)$.
*Этот шаг в проект ещё не добавлен* — здесь мы реализуем его прототипом, потому что он нужен, чтобы увидеть главную картинку CALPHAD:
**касательная к кривой $G_m(x)$** даёт химические потенциалы компонентов.

````python
def n_atoms(model, y):
    return sum(model.site_ratios[s] * y[i] for i, (s, sp) in enumerate(model.species) if sp != 'VA')

def Gm_FeC(x_c, T):
    y_c = x_c / (1 - x_c)                                   # из x_C = y_C / (1 + y_C)
    yv = m.y_from_sublattices([{'FE': 1.0}, {'C': y_c, 'VA': 1 - y_c}])
    return m.gibbs(T, yv) / n_atoms(m, yv)

T = 1200.0
xs = np.linspace(0.001, 0.19, 200)
Gs = np.array([Gm_FeC(x, T) for x in xs])

x0, h = 0.05, 1e-5
G0 = Gm_FeC(x0, T)
slope = (Gm_FeC(x0 + h, T) - Gm_FeC(x0 - h, T)) / (2 * h)
mu_Fe = G0 - x0 * slope                       # пересечение касательной с x = 0
mu_C  = G0 + (1 - x0) * slope                 # пересечение касательной с x = 1

print(f"в точке x(C) = {x0}:  G_m = {G0:.1f} Дж/моль, наклон = {slope:.1f}")
print(f"μ(Fe) = {mu_Fe:.1f} Дж/моль,   μ(C) = {mu_C:.1f} Дж/моль  (относительно референса SER)")

plt.plot(xs, Gs, lw=2.5, label='G_m(x_C), ГЦК Fe–C')
xt = np.array([0.0, 0.19])
plt.plot(xt, mu_Fe + slope * xt, '--', label=f'касательная в x={x0}')
plt.scatter([x0], [G0], color='k', zorder=5)
plt.xlabel('x(C)'); plt.ylabel('G_m, Дж/моль атомов'); plt.title('Касательная = химические потенциалы (1200 К)')
plt.legend(); plt.show()
````

**Условие равновесия двух фаз** — это *общая касательная*: у двух фаз при равновесии совпадают и $\mu_{Fe}$, и $\mu_C$.
Именно это и будет искать солвер (раздел 11).

* аналитический градиент совпадает с конечными разностями на **всех** фазах (масштабированная ошибка ≈ 1e-10).

**Не проверено** (и это важно понимать честно):

* независимая реализация использует **те же соглашения**, что и основной код, — тесты подтверждают, что код реализует формулы, но не что соглашения правильны;
* сравнения с **внешним эталоном** (Thermo-Calc, OpenCalphad, литературные данные) для многокомпонентных составов пока не было — кроме проверки на чистом железе в разделе 11;
* магнитный вклад отсутствует — значения $G$ для магнитных фаз пока неполны.

Два соглашения, которые стоит сверить с эталонным расчётом на нескольких точках: (1) тройной параметр с **единственным** порядком 0 берётся для всех трёх видов
($L_0=L_1=L_2$) — таких троек в базе 124; (2) реципрокный параметр (A,B:C,D) порядка 1 умножается на $(y_A-y_B)$, порядка 2 — на $(y_C-y_D)$; нетривиально это только для `(FE,TI:C,VA)`.

# 11. Что делаем дальше — дорожная карта

Ниже — упорядоченные шаги. У каждого есть вход, результат и критерий готовности (тест).

## Шаг 1. Магнитный вклад (Инден–Хиллерт–Ярл)

Для магнитных фаз (`FCC_A1`, `BCC_A2`, `HCP_A3`, `CEMENTITE`, ...) добавляется

$$G^{mag}=RT\,\ln(\beta+1)\,f(\tau),\qquad \tau=\frac{T}{T_C},$$

$$f(\tau)=\begin{cases}
1-\dfrac{1}{A}\left[\dfrac{79\,\tau^{-1}}{140\,p}+\dfrac{474}{497}\Big(\dfrac1p-1\Big)\Big(\dfrac{\tau^{3}}{6}+\dfrac{\tau^{9}}{135}+\dfrac{\tau^{15}}{600}\Big)\right], & \tau<1\\[3mm]
-\dfrac{1}{A}\left[\dfrac{\tau^{-5}}{10}+\dfrac{\tau^{-15}}{315}+\dfrac{\tau^{-25}}{1500}\right], & \tau\ge 1
\end{cases}
\qquad A=\frac{518}{1125}+\frac{11692}{15975}\Big(\frac1p-1\Big).$$

Здесь $T_C$ и $\beta$ — это **смешанные** `TC` и `BMAGN` фазы: те же суммы «end-members + взаимодействия Редлиха–Кистера», что и для $G$ (класс `InteractionSet` уже умеет это — нужно лишь создать его для типов `TC` и `BMAGN`).
$p$ и AFM-фактор берутся из `TYPE_DEFINITION`. Если $T_C<0$, то $T_C\to T_C/\text{AFM}$; если $\beta<0$, то $\beta\to\beta/\text{AFM}$ (AFM $=-1$ или $-3$) — каждая величина пересчитывается независимо.

**Что нужно сделать:** (а) исправить названия в `TypeDefinition`; (б) добавить `InteractionSet(self, 'TC')` и `'BMAGN'`; (в) реализовать $f(\tau)$ и её производные; (г) учесть нормировку — по моему пониманию, в TDB-модели магнитный вклад на формульную единицу умножается на число атомов $N_{at}$ (для чистых элементов $N_{at}=1$ — различий нет); **это надо сверить с эталонным расчётом**.

Ниже — **прототип** (в проект он ещё не добавлен), проверим его на чистом железе.

````python
def inden_f(tau, p):
    A = 518 / 1125 + 11692 / 15975 * (1 / p - 1)
    if tau < 1:
        return 1 - (79 / (140 * p * tau) + 474 / 497 * (1 / p - 1) * (tau**3 / 6 + tau**9 / 135 + tau**15 / 600)) / A
    return -(tau**-5 / 10 + tau**-15 / 315 + tau**-25 / 1500) / A

def g_mag(T, Tc, beta, p, afm):
    if Tc < 0:                       # условность TDB: отрицательные значения = антиферромагнетик;
        Tc = Tc / afm                #   TC и BMAGN пересчитываются независимо друг от друга
    if beta < 0:
        beta = beta / afm
    if Tc <= 0 or beta <= -1:
        return 0.0
    return R_GAS * T * np.log(beta + 1) * inden_f(T / Tc, p)

# Функция f(tau) должна быть непрерывной в tau = 1
for p_ in (0.4, 0.28):
    print(f"p = {p_}:  f(1−ε) = {inden_f(1 - 1e-9, p_):.9f},  f(1+ε) = {inden_f(1 + 1e-9, p_):.9f}")
````

## Проверка на чистом железе: реальные температуры превращений

В железе есть два превращения: $\alpha$(ОЦК)$\to\gamma$(ГЦК) при ≈1184 К и $\gamma\to\delta$(ОЦК) при ≈1667 К. Они возникают, когда разность энергий
$\Delta(T)=G_{FCC}-G_{BCC}$ меняет знак. Возьмём **все числа из вашей базы** (параметры `G`, `TC`, `BMAGN` и значения $p$, AFM из `TYPE_DEFINITION`)
и добавим магнитный вклад прототипом:

````python
def fe_value(phase, ptype, T):
    return db.evaluate_parameter(db.find_parameter(phase, ptype, [['FE'], ['VA']], 0), T)

def mag_inputs(phase):
    td = db.type_definitions[db.phases[phase].magnetic_symbol]
    afm, p = td.parameters                        # ПЕРВОЕ число — AFM, ВТОРОЕ — p
    return afm, p

def G_fe(phase, T, magnetic=True):
    g = fe_value(phase, 'G', T)
    if magnetic:
        afm, p = mag_inputs(phase)
        g += g_mag(T, fe_value(phase, 'TC', T), fe_value(phase, 'BMAGN', T), p, afm)
    return g

Ts = np.arange(300.0, 2200.0, 0.5)
d_mag = np.array([G_fe('FCC_A1', T) - G_fe('BCC_A2', T) for T in Ts])
d_nomag = np.array([G_fe('FCC_A1', T, False) - G_fe('BCC_A2', T, False) for T in Ts])

def crossings(arr):
    idx = np.where(np.sign(arr[:-1]) != np.sign(arr[1:]))[0]
    return [float(Ts[i]) for i in idx]

print("Нули ΔG = G(FCC) − G(BCC):")
print("   без магнитного вклада:", crossings(d_nomag), "К")
print("   с магнитным вкладом  :", crossings(d_mag), "К     (эксперимент: ≈1184 К и ≈1667 К)")

plt.plot(Ts, d_nomag, label='без магнитного вклада')
plt.plot(Ts, d_mag, lw=2.5, label='с магнитным вкладом (прототип)')
plt.axhline(0, color='k', lw=0.8)
for t in crossings(d_mag):
    plt.axvline(t, color='gray', ls=':'); plt.text(t + 15, 600, f'{t:.0f} К', fontsize=9)
plt.ylim(-400, 800); plt.xlabel('T, К'); plt.ylabel('G(ГЦК) − G(ОЦК), Дж/моль'); plt.legend(loc='lower right')
plt.title('Превращения Fe: α→γ и γ→δ'); plt.show()
````

Это очень сильная проверка: **без** магнитного вклада получаем одно (неверное) пересечение, **с** ним — оба экспериментальных превращения
($\approx 1184$ и $\approx 1667$ К). Она одновременно подтверждает парсеры функций и параметров, обработку `TC`/`BMAGN`, антиферромагнитное соглашение
и — что важно — что в `MAGNETIC -3.0 0.28` первое число действительно AFM, а второе — $p$ (пункт 1 из раздела 10).

Критерий готовности шага 1: этот же тест, но через `PhaseModel` для фазы `BCC_A2`/`FCC_A1` с $y=(\text{Fe}:\text{VA})$;
плюс сравнение градиента конечными разностями и 2–3 сплавных точки с эталонным расчётом.

## Шаг 2. Нормировка и пересчёт составов

Методы `PhaseModel`: `n_atoms(y)`, `mole_fractions(y)` (по элементам, без `VA`), `gibbs_per_atom(T, y)`, и пересчёт производных (раздел 8.5). Критерий: для Fe–C
$x_C = y_C/(1+y_C)$; суммы $x_i$ равны 1; $G$ на атом на вершине `FE:VA` совпадает с $G^{ref}$.

## Шаг 3. Гессиан (вторые производные)

Метод Ньютона для равновесия требует матрицу $\partial^2 G/\partial y_i\,\partial y_j$. Для ideal-части она диагональна ($RT a_s/y_i$), для reference и excess — суммы
произведений по тем же индексным массивам, что уже подготовлены. Критерий: гессиан против конечных разностей **градиента** на всех фазах (по аналогии с тестом градиента).

## Шаг 4. Равновесие внутри одной фазы

Задача: при заданных $T$ и валовом составе $x^0$ найти $\mathbf{y}$, минимизирующий $G_m(\mathbf{y})$ при ограничениях

$$\sum_i y_{s,i}=1\ \ \forall s,\qquad x_i(\mathbf{y})=x_i^0 .$$

Для фаз без внутренних степеней свободы (как `FE:C,VA`) $\mathbf{y}$ однозначно определяется составом. Для фаз с несколькими видами в нескольких подрешётках
(σ, μ, упорядоченные) нужна минимизация методом Лагранжа/Ньютона с барьером или отсечением $y>0$.

## Шаг 5. Равновесие нескольких фаз

Постановка: найти набор фаз, их количества $n^\alpha$ и составы, при которых $G=\sum_\alpha n^\alpha G_m^\alpha$ минимальна. Условия равновесия — **равенство химических потенциалов**
$\mu_i^\alpha=\mu_i$ для всех присутствующих фаз и **неотрицательность движущей силы** для отсутствующих:

$$D^\beta=\min_{\mathbf{y}}\Big[G_m^\beta(\mathbf{y})-\sum_i x_i^\beta(\mathbf{y})\,\mu_i\Big]\;\ge 0 .$$

Схема алгоритма (псевдокод):

```python
def equilibrium(db, T, N):                      # N — мольные количества компонентов
    models = {name: PhaseModel(db, name) for name in candidate_phases}

    # 1. ГЛОБАЛЬНЫЙ поиск стартовой точки: сетка/случайные y в каждой фазе,
    #    выпуклая оболочка (или линейное программирование) по (x, G_m)
    points = sample_all_phases(models, T)
    phase_set, start = lower_convex_hull(points, N)

    while True:
        # 2. ЛОКАЛЬНОЕ уточнение: метод Ньютона на системе
        #    mu_i^a = mu_i,  sum_a n^a x_i^a = N_i,  ограничения подрешёток
        state = newton_refine(models, T, N, phase_set, start)

        # 3. ПРОВЕРКА устойчивости: нет ли фазы, которая снизила бы G
        D, y_new, phase_new = most_negative_driving_force(models, T, state.mu)
        if D >= -tol:
            break
        phase_set.add(phase_new, y_new)                    # добавить фазу с отрицательным D

        # 4. Если количество фазы стало < 0, фаза удаляется из набора
        phase_set = drop_phases_with_negative_amount(state)

    return state                                           # фазы, количества, составы, mu
```

Именно здесь нужны: нормировка (шаг 2), градиент (есть), гессиан (шаг 3), отсутствующие end-members (пункт 2 раздела 10).

## Шаг 6. Валидация солвера

* Fe–C: эвтектоид в метастабильной системе Fe–Fe₃C около 727 °C (≈1000 К) — известное значение, с которым стоит сравнить;
* температура превращений и растворимость C в феррите/аустените;
* сравнение с Thermo-Calc/OpenCalphad на наборе сплавных точек (Fe–Cr–Ni, Fe–C–Mn);
* регрессионные тесты: любое изменение кода проверяется на фиксированном наборе точек.

## Шаг 7. Скорость и упаковка

Сейчас $G$+градиент для типичной фазы — порядка 0.1–0.3 мс. Для карты равновесий нужны тысячи вызовов, поэтому позже имеет смысл:
кэшировать значения параметров по $T$ (уже сделано), векторизовать по нескольким $y$ сразу, при необходимости — `numba`.
Упаковка: класс `Equilibrium(db).calculate(T, composition)` с понятным результатом.

## Шаг 8. Разупорядоченные/упорядоченные фазы (`BCC_B2`) и composition sets

Судя по закомментированной команде `ATTACH_CONTRIBUTION BCC_B2 BCC_A2 ORDER_DISORDER`, `BCC_B2` в оригинале — упорядоченная часть **модели с разделением энергии**
$G = G^{dis}(x) + G^{ord}(\mathbf{y}) - G^{ord}(\mathbf{y}=x)$, где $G^{dis}$ берётся из `BCC_A2`. Вероятно, именно поэтому у `BCC_B2` лишь 33 end-members из 2400:
определена только упорядоченная добавка (это гипотеза по комментариям скрипта очистки; её надо проверить по исходной базе). Нужно вернуть эту связь (ATTACH_CONTRIBUTION в исходной базе) и вместе с ней — возможность иметь несколько composition sets одной фазы
(`ADD_COMPOSITION_SET`) для разрывов растворимости.

## Глоссарий

| Термин | Смысл |
|---|---|
| CALPHAD | метод расчёта фазовых равновесий по моделям энергии Гиббса фаз |
| TDB | текстовый формат базы данных (команды ELEMENT, PHASE, PARAMETER, ...) |
| SER | Standard Element Reference — стандартное состояние элемента (`GHSERFE`) |
| CEF / подрешёточная модель | Compound Energy Formalism — фаза описывается подрешётками |
| site fraction $y_{s,i}$ | доля вида $i$ в подрешётке $s$ |
| end-member | «чистое соединение»: по одному виду в каждой подрешётке |
| `VA` | вакансия — вид, не являющийся атомом |
| Redlich–Kister | полиномиальное разложение энергии бинарного взаимодействия по $(y_i-y_j)^v$ |
| Muggianu | способ продолжения бинарных моделей на тройные системы |
| химический потенциал $\mu_i$ | $\partial G/\partial N_i$ — «давление» компонента; в равновесии одинаков во всех фазах |
| движущая сила $D$ | насколько может снизиться $G$ при появлении новой фазы |
| composition set | один из нескольких «экземпляров» одной фазы (разрыв растворимости) |
| $T_C$, $\beta$, $p$ | температура Кюри/Нееля, магнитный момент, структурный фактор модели Индена |
